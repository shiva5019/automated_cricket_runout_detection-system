# 🏏 Cricket Run-Out Detection using Deep Learning
### Binary Image Classification — OUT vs NOT OUT using MobileNetV2 and ResNet50V2

```python
import os
import random
import shutil
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2, ResNet50V2
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras import backend as K
from sklearn.metrics import classification_report, confusion_matrix
from PIL import Image as PILImage
import seaborn as sns

print("TensorFlow Version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))
```

```python
## Configuring Paths & Hyperparameters
```

```python
BASE_DIR    = r"C:\Users\J.Shiva\OneDrive\Attachments\cricket_runout_detection"
DATA_DIR    = os.path.join(BASE_DIR, "data", "train")
SPLIT_DIR   = os.path.join(BASE_DIR, "data", "split")
RESULTS_DIR = os.path.join(BASE_DIR, "results")
MODELS_DIR  = os.path.join(BASE_DIR, "models")

MOBILENET_SAVE_PATH = os.path.join(MODELS_DIR, "best_model.keras")
RESNET_SAVE_PATH    = os.path.join(MODELS_DIR, "best_model_resnet.keras")

CLASSES    = ["out", "not_out"]
IMG_SIZE   = (224, 224)
BATCH_SIZE = 16
EPOCHS     = 20
LR         = 0.0001

os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(MODELS_DIR,  exist_ok=True)

print("Base dir   :", BASE_DIR)
print("Split dir  :", SPLIT_DIR)
print("Image size :", IMG_SIZE)
print("Batch size :", BATCH_SIZE)
print("Epochs     :", EPOCHS)
print("LR         :", LR)
```

```python
## Splitting Dataset into Train / Val / Test
```

```python
random.seed(42)

for cls in CLASSES:
    src_folder = os.path.join(DATA_DIR, cls)
    images     = os.listdir(src_folder)
    random.shuffle(images)

    n_total = len(images)
    n_test  = int(n_total * 0.15)
    n_val   = int(n_total * 0.15)

    test_imgs  = images[:n_test]
    val_imgs   = images[n_test:n_test + n_val]
    train_imgs = images[n_test + n_val:]

    for split, split_imgs in [("train", train_imgs), ("val", val_imgs), ("test", test_imgs)]:
        dest = os.path.join(SPLIT_DIR, split, cls)
        os.makedirs(dest, exist_ok=True)
        for img in split_imgs:
            shutil.copy(os.path.join(src_folder, img), os.path.join(dest, img))

print("Split Complete!")
for split in ["train", "val", "test"]:
    print(f"\n{split}/")
    for cls in CLASSES:
        count = len(os.listdir(os.path.join(SPLIT_DIR, split, cls)))
        print(f"  {cls}: {count} images")
```

```python
## Creating Data Generators
```

```python
TRAIN_DIR = os.path.join(SPLIT_DIR, "train")
VAL_DIR   = os.path.join(SPLIT_DIR, "val")
TEST_DIR  = os.path.join(SPLIT_DIR, "test")

datagen = ImageDataGenerator(rescale=1./255)

train_gen = datagen.flow_from_directory(
    TRAIN_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode="binary"
)
val_gen = datagen.flow_from_directory(
    VAL_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode="binary"
)
test_gen = datagen.flow_from_directory(
    TEST_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode="binary", shuffle=False
)

print("Class indices:", train_gen.class_indices)
print("Train batches:", len(train_gen))
print("Val batches  :", len(val_gen))
print("Test batches :", len(test_gen))
```

```python
## Building MobileNetV2 Model
```

```python
base_model = MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights="imagenet")
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dropout(0.3)(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.2)(x)
output = Dense(1, activation="sigmoid")(x)

model = Model(inputs=base_model.input, outputs=output)

model.compile(
    optimizer=Adam(learning_rate=LR),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("Trainable params    :", sum([tf.size(w).numpy() for w in model.trainable_weights]))
print("Non-trainable params:", sum([tf.size(w).numpy() for w in model.non_trainable_weights]))
```

```python
## Loading Saved MobileNetV2 Model
```

```python
model = tf.keras.models.load_model(MOBILENET_SAVE_PATH)
print("MobileNetV2 loaded successfully")
```

```python
## Training MobileNetV2 Model
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau

MODEL_SAVE_PATH = r"C:\Users\J.Shiva\OneDrive\Attachments\cricket_runout_detection\models\best_model.keras"

callbacks = [
    EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True, verbose=1),
    ModelCheckpoint(MODEL_SAVE_PATH, monitor="val_accuracy", save_best_only=True, verbose=1),
    ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-7, verbose=1)
]

history = model.fit(
    train_gen,
    epochs=EPOCHS,
    validation_data=val_gen,
    callbacks=callbacks
)
```

```python
## Plotting MobileNetV2 Training Curves
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(history.history["accuracy"], label="Train Accuracy")
axes[0].plot(history.history["val_accuracy"], label="Val Accuracy")
axes[0].set_title("Accuracy over Epochs")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].legend()
axes[0].grid(True)

axes[1].plot(history.history["loss"], label="Train Loss")
axes[1].plot(history.history["val_loss"], label="Val Loss")
axes[1].set_title("Loss over Epochs")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.savefig(r"C:\Users\J.Shiva\OneDrive\Attachments\cricket_runout_detection\results\training_curves.png", dpi=150)
plt.show()
```

```python
## Evaluating MobileNetV2 on Test Set
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

test_gen.reset()
y_pred_probs = model.predict(test_gen, verbose=1)
y_pred = (y_pred_probs > 0.5).astype(int).flatten()
y_true = test_gen.classes

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=["not_out", "out"]))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["not_out", "out"],
            yticklabels=["not_out", "out"])
plt.title("Confusion Matrix — Test Set")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.savefig(r"C:\Users\J.Shiva\OneDrive\Attachments\cricket_runout_detection\results\confusion_matrix.png", dpi=150)
plt.show()
```

```python
## Building ResNet50V2 Model
```

```python
from tensorflow.keras.applications import ResNet50V2
from tensorflow.keras import backend as K

K.clear_session()

base_model_resnet = ResNet50V2(input_shape=(224, 224, 3), include_top=False, weights="imagenet")
base_model_resnet.trainable = False

x = base_model_resnet.output
x = GlobalAveragePooling2D()(x)
x = Dropout(0.3)(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.2)(x)
output = Dense(1, activation="sigmoid")(x)

model_resnet = Model(inputs=base_model_resnet.input, outputs=output)

model_resnet.compile(
    optimizer=Adam(learning_rate=LR),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("Trainable params    :", sum([tf.size(w).numpy() for w in model_resnet.trainable_weights]))
print("Non-trainable params:", sum([tf.size(w).numpy() for w in model_resnet.non_trainable_weights]))
```

```python
## Training ResNet50V2 Model
```

```python
RESNET_SAVE_PATH = r"C:\Users\J.Shiva\OneDrive\Attachments\cricket_runout_detection\models\best_model_resnet.keras"

callbacks_resnet = [
    EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True, verbose=1),
    ModelCheckpoint(RESNET_SAVE_PATH, monitor="val_accuracy", save_best_only=True, verbose=1),
    ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-7, verbose=1)
]

history_resnet = model_resnet.fit(
    train_gen,
    epochs=EPOCHS,
    validation_data=val_gen,
    callbacks=callbacks_resnet
)
```

```python
## Plotting ResNet50V2 Training Curves
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(history_resnet.history["accuracy"], label="Train Accuracy")
axes[0].plot(history_resnet.history["val_accuracy"], label="Val Accuracy")
axes[0].set_title("ResNet50V2 — Accuracy over Epochs")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].legend()
axes[0].grid(True)
axes[1].plot(history_resnet.history["loss"], label="Train Loss")
axes[1].plot(history_resnet.history["val_loss"], label="Val Loss")
axes[1].set_title("ResNet50V2 — Loss over Epochs")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].legend()
axes[1].grid(True)
plt.tight_layout()
plt.savefig(r"C:\Users\J.Shiva\OneDrive\Attachments\cricket_runout_detection\results\resnet_training_curves.png", dpi=150)
plt.show()
```

```python
## Evaluating ResNet50V2 on Test Set
```

```python
test_gen.reset()
y_pred_probs_resnet = model_resnet.predict(test_gen, verbose=1)
y_pred_resnet = (y_pred_probs_resnet > 0.5).astype(int).flatten()
y_true = test_gen.classes

print("\nClassification Report — ResNet50V2:")
print(classification_report(y_true, y_pred_resnet, target_names=["not_out", "out"]))

cm_resnet = confusion_matrix(y_true, y_pred_resnet)
plt.figure(figsize=(6, 5))
sns.heatmap(cm_resnet, annot=True, fmt="d", cmap="Greens",
            xticklabels=["not_out", "out"],
            yticklabels=["not_out", "out"])
plt.title("Confusion Matrix — ResNet50V2 Test Set")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.savefig(r"C:\Users\J.Shiva\OneDrive\Attachments\cricket_runout_detection\results\resnet_confusion_matrix.png", dpi=150)
plt.show()
```
